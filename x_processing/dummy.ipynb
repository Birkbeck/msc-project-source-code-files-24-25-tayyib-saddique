```python
import psutil

print(psutil.cpu_count(logical=True))   # total logical cores
print(psutil.cpu_count(logical=False))  # physical cores
print(psutil.cpu_percent(interval=1, percpu=True))  # CPU usage per core
```

```python
# List top CPU-consuming processes
for proc in psutil.process_iter(['pid', 'name', 'cpu_percent']):
    try:
        if proc.info['cpu_percent'] > 1:  # threshold
            print(proc.info)
    except psutil.NoSuchProcess:
        pass
```
