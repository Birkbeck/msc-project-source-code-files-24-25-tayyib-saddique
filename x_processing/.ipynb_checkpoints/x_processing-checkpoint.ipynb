```python
from transformers import pipeline
import time

model = pipeline('text-classification', model = 'finiteautomata/bertweet-base-sentiment-analysis', device=0)

texts = ['This is a polticial statement.']*50000
start = time.time()
preds = model(texts, batch_size=128)
end = time.time()

print(f'Processed 1000 tweets in {end - start:.2f} seconds')
```
