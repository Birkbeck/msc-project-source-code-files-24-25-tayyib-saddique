```python
import pandas as pd
import pyarrow.parquet as pq
import pyarrow.compute as pc
import joblib
import os
```

```python
UNLABELED_PARQUET = "train_unlabelled.parquet"

table = pq.read_table("train_unlabelled.parquet", columns=[])
print(f"Number of rows: {table.num_rows:,}")
```

```python
LABELLED_PARQUET = "train_labelled.parquet"

table = pq.read_table(LABELLED_PARQUET, columns=["party", "sentiment"])

labelled_count = table.num_rows

party_counts = {
    str(row["values"].as_py()): row["counts"].as_py()
    for row in pc.value_counts(table["party"])
}

sentiment_by_party = {}
parties = table["party"].unique().to_pylist()
for party in parties:
    mask = pc.equal(table["party"], party) 
    subset = table.filter(mask)

    counts = pc.value_counts(subset["sentiment"])
    sentiment_by_party[party] = {
        str(row["values"].as_py()): row["counts"].as_py()
        for row in counts
    }

print(f"\nLabelled samples: {labelled_count:,}")
print(f"Party distribution: {party_counts}")
for party, dist in sentiment_by_party.items():
    print(f"  {party}: {dist}")
```

```python
MODEL_DIR = "models"
UNLABELED_PARQUET = "train_unlabelled.parquet"
OUTPUT_PARQUET = "predictions.parquet"

def load_model(classifier, party=None, directory=MODEL_DIR):
    """
    Dynamically load the most recent model for the given classifier.
    
    classifier: "party" or "sentiment"
    party: required for sentiment models ("democrat" or "republican")
    """
    candidates = []

    for f in os.listdir(directory):
        if not f.endswith(".joblib"):
            continue
        f_lower = f.lower()
        if classifier == "party" and "party_classifier" in f_lower:
            candidates.append(f)
        elif classifier == "sentiment" and party and party.lower() in f_lower and "sentiment_classifier" in f_lower:
            candidates.append(f)

    if not candidates:
        raise FileNotFoundError(f"No model found for classifier='{classifier}' party='{party}'")

    model_path = os.path.join(directory, candidates[0])

    print(f"Loading {classifier} model from: {model_path}")
    return joblib.load(model_path)

party_model = load_model("party")
dem_sentiment_model = load_model("sentiment", party="democrat")
rep_sentiment_model = load_model("sentiment", party="republican")
```

```python
unlabeled_df['predicted_party'] = party_model.predict(unlabeled_df['clean_text'])

def predict_sentiment(row):
    if row['predicted_party'] == 'democrat':
        return dem_sentiment_model.predict([row['clean_text']])[0]
    elif row['predicted_party'] == 'republican':
        return rep_sentiment_model.predict([row['clean_text']])[0]
    else:
        return None

unlabeled_df['predicted_sentiment'] = unlabeled_df.apply(predict_sentiment, axis=1)

unlabeled_df.to_parquet(OUTPUT_PARQUET, index=False)
print(f"Predictions saved to {OUTPUT_PARQUET}")
```
